Q15. Menu-driven calculator

```python
num1 = float(input("Enter the 1st number:"))
num2 = float(input("Enter the 2nd number:"))
symbol = input('Enter the operator (+, -, *, /, //, %, **) :')
if symbol=='+':
    result = num1+num2
    print(f'The result is {result}')
elif symbol=='-':
    result = num1-num2
    print(f'The result is {result}')
elif symbol=='*':
    result = num1*num2
    print(f'The result is {result}')
elif symbol=='/':
    if num2==0:
        print('Division by zero is not allowed')
    else:
        print('The result is :', num1/num2)
elif symbol=='//':
    if num2==0:
        print('The floor division by zero is not allowed')
    else:
        print('The result is :', num1//num2)
elif symbol=='%':
    if num2==0:
        print('The modulo division by zero is not allowed')
    else:
        print("the result is :", num1%num2)
elif symbol=='**':
    result = num1**num2
    print(f'The result is {result}')
else:
    print('Invalid Operator')



```

Q22. Angles form triangle

```python
angle1 = int(input("Enter the 1st angle: "))
angle2 = int(input("Enter the 2nd angle: "))
angle3 = int(input("Enter the 3rd angle: "))
if angle1 <= 0 or angle2 <= 0 or angle3 <= 0:
    print("Invalid Input")
else:
    if angle1 + angle2 + angle3 != 180:
        print("Invalid Triangle")
    else:
        if angle1 == 90 or angle2 == 90 or angle3 == 90:
            print("Right Angle Triangle")
        elif angle1 > 90 or angle2 > 90 or angle3 > 90:
            print("Obtuse Triangle")
        else:
            print("Acute Triangle")
```
