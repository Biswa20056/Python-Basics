```python
num = 4
spaces = num-1
stars = 1
for row in range(1,num+1):
    for col1 in range(1,stars+1):
        print('*',end=' ')
    for col2 in range(1,spaces+1):
        print(' ',end=' ')
    print()
    stars+=1
    spaces-=1

```
