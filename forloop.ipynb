Q5. Wap to print factorail of a number.

```python
n = int(input("Enter the number:"))
fact = 1
if n==0:
    print(f'The factorial of {n} is 1.')
elif n<0:
    print(f'The factorial of negative {n} is not possible.')
else:
    for num in range(1,n+1):
        fact*=num
    print(f'The factorial of {n} is {fact}.')
```

Q6. Wap to print fizz if num is div by 2, print buzz if num is div by 3 and print fizzbuzz if num is div by 3 and 5 in a range.

Q7. Wap to print the sum of digits of a number.

```python
n = abs(int(input("Enter the number:")))
total = 0
while n>0:
    remainder = n%10
    total += remainder
    n //= 10
print(f'The sum of digits is {total}.')
```

Q8. Wap to print a number is a perfect number or not.

```python
n = int(input("Enter the number:"))
total = 0
for num in range(1,n):
    if (n%num==0):
        total += num
if (n==total):
    print(f'{n} is a perfect number.')
else:
    print(f'{n} is not a perfect number.')
```

Q15. WAP TO PRINT A NUMBER IS A SPY NUMBER OR NOT(THE SUM OF DIGITS == THE PRODUCT OF DIGITS)

```python
n = 123
temp = n
total = 0
Multi = 1
while temp>0:
    rem = temp%10
    total+=rem
    Multi*=rem
    temp//=10
if (total==Multi):
    print(f"{n} is a Spy number.")
else:
    print(f"{n} is not Spy number.")

```

Q16. Wap to print the sum of the square of the digits of a number.

```python
num = int(input("Enter the number:"))
temp = num
res = 0
while temp>0:
    rem = temp%10
    res = res + rem**2
    temp//=10
print(f'The result is: {res}')
```

Q17. Wap to print a number is a Amstrong number or not.

```python
n = int(input("Enter the number:"))
dup = n
digits = len(str(n))
res = 0
while n>0:
    rem = n%10
    res = res + (rem ** digits)
    n//=10
if (dup==res):
    print("Amstrong Number.")
else:
    print("Not Amstrong Number.")
```
