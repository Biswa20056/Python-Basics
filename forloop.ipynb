Q17. Wap to print a number is a Amstrong number or not.

```python
n = int(input("Enter the number:"))
dup = n
digits = len(str(n))
res = 0
while n>0:
    rem = n%10
    res = res + (rem ** digits)
    n//=10
if (dup==res):
    print("Amstrong Number.")
else:
    print("Not Amstrong Number.")
```

Q17. Wap to check a number is dissaroum number or not
