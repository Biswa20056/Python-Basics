1. count():It returns the occurrences of given substring and it returns integer.

Syntax1: variablename.count(substring)

```python
s = 'abcdea'
print(s.count('a'))# it will give the frequency of substring in the main string
print(s.count('j'))# if substring is not present it will simply return 0
print(s.count('ab'))
print(s.count(''))# it will return len+1 bcoz after and before substring there is an empty string
```

```python
s = 'abcdabcded'
print(s.count('a',0))# from 0th index to last how many a is there
print(s.count('a',1))
print(s.count('d',-3))# no matter what but the operation will be performed from left to right direction
print(s.count('ab',0))
print(s.count('ab',1))
print(s.count('cd',2))
print(s.count('cd',3))
print(s.count('cd',4))
print(s.count('h',2))
```
