Built-in functions

String-Built-in Functions:

1. count():It returns the occurrences of given substring and it returns integer.

Syntax1: variablename.count(substring)

```python
s = 'abcdea'
print(s.count('a'))# it will give the frequency of substring in the main string
print(s.count('j'))# if substring is not present it will simply return 0
print(s.count('ab'))
print(s.count(''))# it will return len+1 bcoz after and before substring there is an empty string
```

Syntax2: variablename.count(substring,starting position)

```python
s = 'abcdabcded'
print(s.count('a',0))# from 0th index to last how many a is there
print(s.count('a',1))
print(s.count('d',-3))# no matter what but the operation will be performed from left to right direction
print(s.count('ab',0))
print(s.count('ab',1))
print(s.count('cd',2))
print(s.count('cd',3))
print(s.count('cd',4))
print(s.count('h',2))
```

Syntax3: variablename.count(substring, starting position, ending position)

```python
s = 'abcdbcdabc'
print(s.count('a',2,7))# this syntax will not include the ending position
print(s.count('a',2,8))
print(s.count('ab',0,3))
print(s.count('c',-5,0))
print(s.count('c',-5,-3))
print(s.count('c',-5,8))
print(s.count('k',0,len(s)))
```

2. Split Function:

1. split function breaks the string based on given substring
2. if no argument is given space is default substring
3. split function returns list of substrings.

Syntax1: variablename.split(substring)

```python
s = 'Hello bye world'
print(s.split('e'))# the number of times split happens more than one will be output in list format
print(s.split())
print(s.split('l'))
print(s.split('ll'))
print(s.split('k'))
```

Syntax2: variablename.split()

```python
s = 'hello bye world'
print(s.split())
print(s.split(' '))
```

Syntax3: variablename.split(substring,count)
 Here count refers to how many times you want to split

Note: After one split and before one split all the characters should be there if there is no substring then empty string will takes place

```python
s = 'abcbacabc'
print(s.split('b'))
print(s.split('b',1))
print(s.split('b',2))
print(s.split('b',10))# the max number of b will be there as many times split will happend and rest will be ignored
```

3. index function:
1. Index function return the index position of the first occurence of that substring
2. if given subnstring is not present then index function throws valueError
3. index function returns integer

Syntax1: variablename.index(substring)

```python
s = 'abcdabcdabcd'
print(s.index('a'))
print(s.index('d'))
# it will raise value error print(s.index('y'))
# print(s.index()) atleast pne argument should be inside index function
```

Syntax2: variablename.index(substring, starting position)

```python
s = 'abcdabcdabcd'
print(s.index('c',2))
print(s.index('c',3))
print(s.index('c',7))
print(s.index('b',-9))
# print(s.index('c',11) this will give substring out of range error
```

Syntax3: variablename.index(substring,starting position, ending position)

```python
s = 'abcabcabc'
print(s.index('c',1,11))
print(s.index('c',-7,-3))
#print(s.index('c',-7,1)) this will also raise error that substring not found
print(s.index('c',-8,10))
```

to get the second occurence of any substring

Syntax: varname.index(substring, varname.index(substring)+1)

```python
s = 'abcabcabc'
print(s.index('a'))# 1st occurrence
print(s.index('a', s.index('a')+1))# 2nd occurrence
print(s.index('a', s.index('a', s.index('a')+1)+1))# 3rd occurrence
```

4. find function:

1. Find function return the index position of first occurence of substring
2. if the given substring is not present it simply returns -1
3. find function returns integer
4. all the 3 syntaxes applicable for index function also be for find function


Syntax1: varname.find(substring)

```python
s = 'abcabcabc'
print(s.find('a'))
print(s.find('b'))
print(s.find('k'))# this substr is not present so it returned -1
```

Syntax2: varname.find(substring, starting pos)

```python
s = 'abcabcabc'
print(s.find('b',1))
print(s.find('c',-4))
print(s.find('p',4))
print(s.find('a',1))
```

Syntax3: varname.find(substring, starting pos, ending pos)

```python
s = 'abcabcabc'
print(s.find('a',0,2))
print(s.find('a',1,3))
print(s.find('b',-3,6))
print(s.find('b',-3,len(s)))
```

5. rindex function:

1. rindex function returns the last occurrence of the substring in agive string
2. in every searching will perform from left to right of the main string
3. if the substring is not present in that main string rindex will throw error
4. rindex returns integer

Syntax1: varname.rindex(substring)

```python
s = 'abcdabcdabcd'
print(s.rindex('a'))
print(s.rindex('d'))
print(s.rindex('c'))
# print(s.rindex('p')) this will raise value error
```

syntax2: varname(substring, starting pos)

```python
s = 'abcdabcdabcd'
print(s.rindex('b',1))
print(s.rindex('a',1))
print(s.rindex('d',10))
```

syntax3: varname.rindex(substring, starting pos, ending position)

```python
s = 'abcdabcdabcd'
# this will raise error bcoz in that range no substring 'a' is present print(s.rindex('a',1,4))
print(s.rindex('a',0,4))
print(s.rindex('b',1,8))
```

To get last but one occurrence means the second occurrence from last direction of a substring by rindex

```python
s = 'abcdabcdabcd'
print(s.rindex('a'))# last occurence
print(s.rindex('a', 0, s.rindex('a')))# last but one
print(s.rindex('b'))
print(s.rindex('b',0,s.rindex('b')))# last but one index
print(s.rindex('b', 0, s.rindex('b',0,s.rindex('b'))))# last but 2 index
```

6. rfind function:

1. rfind functions returns the last occcurence of a subtring
2. if the given substring is not present it will simply return -1
3. r find function returns integer

Syntax1: varname.rfind(substring)

```python
s = 'abcdabcdabcd'
print(s.rfind('a'))
print(s.rfind('d'))
print(s.rfind('p'))
```

Syntax2: varname.rfind(substring, starting position)

```python
s = 'abcdabcdabcd'
print(s.rfind('a',0))
print(s.rfind('d',12))
print(s.rfind('p',4))
print(s.rfind('d',-7))
```

syntax3: varname.rfind(substring, starting position, ending position)

```python
s = 'abcdabcdabcd'
print(s.rfind('a',0,3))
print(s.rfind('a',1,3))
print(s.rfind('b',2,4))
print(s.rfind('d',-3,len(s)))
print(s.rfind('c',-6,-2))
print(s.rfind('c',-6,-1))
```

to get the last but one and last but two index through rfind function

```python
s = 'abcdabcdabcd'
print(s.rfind('a'))
print(s.rfind('a',0,s.rfind('a')))# last but one
print(s.rfind('a',0,s.rfind('a',0,s.rfind('a'))))# last but two
print(s.rfind('a',0,s.rfind('a',0,s.rfind('a',0,s.rfind('a')))))
```

7. replace function:

1. replace function chnages the old substring with the new substring in new memory location.
2. Existing memory location will not be change it will be constant or remain same.
3. replace function returns string(with new memory location)
4. generaaly replace function creates a new memory location.

Syntax1: varnname.replace(old substring, new ubstring)

```python
s = 'abcdacd'
print(s.replace('a','A'))
print(s.replace('D','d'))
# print(s.replace()) it will raise error bcoz replace function takes atleast 2 argument
print(s.replace('a',''))# this will delete that substring 
print(s.replace('y','a'))# if substr is no present it will leave it as it is
print(s.replace('bc','hell'))
```

Syntax2: varname.replace(old substr, new substr, count) Here count refers to how many times you want to replace

```python
s = 'abcdacD'
print(s.replace('a','A',1))
print(s.replace('a','A',2))
print(s.replace('a','A',10)) # maxm times replace will happen and rest will be ignored
print(s.replace('a','',1))
print(s.replace('D','',3))
```
