16. Count primes

```python
n = int(input("Enter the number :"))
is_prime = [True]*n
is_prime[0] = is_prime[1] = False
p = 2
while p*p<n:
    if is_prime[p]:
        for multiple in range(p*p,n,p):
            is_prime[multiple]=False
    p+=1
print(sum(is_prime))
```

38. Decimal to hexadecimal

```python
n = int(input("Enter the number : "))
dup = n
place = 1
hexa = ""
while n > 0:
    rem = n % 16
    if rem < 10:
        hexa = str(rem) + hexa
    elif rem == 10:
        hexa = "A" + hexa
    elif rem == 11:
        hexa = "B" + hexa
    elif rem == 12:
        hexa = "C" + hexa
    elif rem == 13:
        hexa = "D" + hexa
    elif rem == 14:
        hexa = "E" + hexa
    else:
        hexa = "F" + hexa
    n //= 16
print(hexa)
```

46. Number pyramid

```python
n = 5
spaces = n-1
stars = 1
for row in range(1,n+1):
    for col1 in range(1,spaces+1):
        print(' ',end=' ')
    data = 1
    for col2 in range(1,stars+1):
        print(data,end=' ')
        if col2<row:
            data+=1
        else:
            data-=1
    print()
    spaces-=1
    stars+=2
```

51. butterfly pattern

```python
num = 5
spaces = (num-1)*2
stars = 1
for row in range(1,num*2):
    for col1 in range(1,stars+1):
        print(col1,end=' ')
    for col2 in range(1,spaces+1):
        print(' ',end=' ')
    for col3 in range(1,stars+1):
        print(col3,end=' ')
    print()
    if row<num:
        spaces-=2
        stars+=1
    else:
        spaces+=2
        stars-=1
```
