1. Print 1 to N

```python
n = 1
while n<=10:
    print(n)
    n+=1
```

2. Print N to 1

```python
n = 10
while n>=1:
    print(n)
    n-=1
```

3. Sum of digits

```python
n = 890
dup = n
total = 0
while n>0:
    rem = n%10
    total += rem
    n//=10
print(f'The sum of digits of {dup} is : {total}')
```

4. Reverse number

```python
n = 987
dup = n
rev = 0
while n>0:
    rem = n%10
    rev = rev*10 + rem
    n//=10
print(f'The reverse of {dup} is : {rev}')
```

5. Palindrome

```python
n = 121
dup = n
rev = 0
while n>0:
    rem = n%10
    rev = rev*10 + rem
    n//=10
if (dup==rev):
    print(f'The {dup} is a pallindrome number')
else:print(f'The {dup} is not a pallindrome number')
```

6. Armstrong

```python
n = 9474
dup = n
digits = len(str(n))
total = 0
while n>0:
    rem = n%10
    total = total + rem**digits
    n//=10
if (dup==total):
    print(f'The {dup} is a amstrong number')
else:
    print(f'The {dup} is not a amstrong number')
```

7. Strong

```python
n = 145
dup = n
res = 0
while n>0:
    rem = n%10
    fact = 1
    while rem>0:
        fact*=rem
        rem-=1
    res += fact
    n//=10
if dup==res:
    print(f'{dup} is a strong number')
else:print(f'{dup} is not a strong number')

    
```

8. Perfect number

```python
n = 9
dup = n
i = 1
factors = 0
while i<n//2+1:
    if n%i==0:
        factors +=i
    i+=1
if dup==factors:
    print(f'{dup} is a Perfect number')
else:
    print(f'{dup} is not a perfect number')
```

9. Fibonacci

```python
n = 5
a = 0
b = 1
i = 1
while i<=n:
    print(a,end=' ')
    a,b = b, a+b
    i+=1
print()
```
