1. largest number

```python
a = [3,2,1,5,2]
n = len(a)
largest = a[0]
for idx in range(n):
    if a[idx]>largest:
        largest = a[idx]
print(largest)
```

second largest

```python
a = [3,2,1,5,2]
n = len(a)
largest = a[0]
S_largest = -1
for idx in range(1,n):
    if a[idx]>largest and largest != S_largest:
        S_largest = largest
        largest = a[idx]
    else:
        if a[idx]<largest and a[idx]>S_largest:
            S_largest = a[idx]
print(S_largest)
```

Smallest

```python
a = [3,2,1,5,2]
n = len(a)
smallest = a[0]
for idx in range(n):
    if a[idx]<smallest:
        smallest = a[idx]
print(smallest)
```

Second smallest

```python
a = [3,2,1,5,2]
n = len(a)
smallest = a[0]
S_smallest = float('inf')
for idx in range(1,n):
    if a[idx]<smallest:
        S_smallest = smallest
        smallest = a[idx]
    else:
        if a[idx]!=smallest and a[idx]<S_smallest:
            S_smallest = a[idx]
print(S_smallest)
```

check array is sorted or not

```python
a = [1,2,2,1,3,4]
n = len(a)
for idx in range(n-1):
    if a[idx+1]<a[idx]:
        print('False')
        break
else:
    print('True')
```

count unique element in an array

```python
a = [1,1,2,2,3,3,4]
n = len(a)
i = 0
for j in range(1,n):
    if a[i]!=a[j]:
        a[i+1]=a[j]
        i+=1
print(i+1)
```

reverse an array

```python
a = [1,2,3,4,5]
n = len(a)
i = 0
j = n-1
while i<j:
    a[i],a[j] = a[j],a[i]
    i+=1
    j-=1
print(a)
```

Search ana element

```python
a = [10,20,30,40,50]
n = len(a)
target = 900
found = -1
for idx in range(n):
    if a[idx]==target:
        found = idx
        break
print(found)
```

left rotate by one place

```python
a = [1,2,3,4,5]
n = len(a)
k = 1
temp = a[0]
for idx in range(k,n):
    a[idx-1] = a[idx]
a[n-1] = temp
print(a)
```

rotate by k place

```python
a = [1,2,3,4,5,6]
n = len(a)
k = 3
temp = a[:k]
for i in range(k,n):
    a[i-k] = a[i]
j = 0
for i in range(n-k,n):
    a[i] = temp[j]
    j+=1
print(a)
```
