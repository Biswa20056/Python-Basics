```python
a = [1,2,3,5,6,7]
b =[1,2,4,5,6]
n1 = len(a)
n2 = len(b)
ans = []
i = 0
j = 0
while i<n1 and j<n2:
    if a[i]<b[j]:
        i+=1
    elif b[j]<a[i]:
        j+=1
    else:
        ans.append(a[i])
        i+=1
        j+=1
print(ans)
```
