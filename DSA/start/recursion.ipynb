Recursion

When a function calls itself

Head recursion means first the job is done and then function call happend

```python
def name(x):
    if x==4:
        return
    print('Biswajit')
    x+=1
    name(x)

x = 0
name(x)

```

Taill recursion means first function call then job gets done

```python
def name(x):
    if x==4:
        return
    x+=1
    name(x)
    print('Biswajit')
name(x)
x = 0
```

Functional recursion

```python
def sum(x,res):
    if x==6:
        print(res)
        return
    sum(x+1,res+x) 
    

x = 1
res = 0
sum(x,res)
```

```python
def func(n):
    if n==1:
        return 1
    return n + func(n-1)

n = 10
func(n)
```

factorial

```python
def Factorial(n):
    if n==0:
        return 1
    return n * Factorial(n-1)
n = 5
Factorial(n) 
```

reverse an array using recursion

```python
a = [5,7,3,2,6,1,5,9]
i = 0
j = len(a)-1
while i<j:
    a[i],a[j] = a[j],a[i]
    i+=1
    j-=1
print(a)
```

reverse a specific part of an array

```python
a = [5,7,3,2,6,1,5,9]
left = 2
right = 5# that means reverse from second index to 5th index
i = left
j = right
while i<j:
    a[i],a[j] = a[j],a[i]
    i+=1
    j-=1
print(a)
```

reverse by recursion

```python
def rev(a,i,j):
    if i>=j:
        return a
    a[i],a[j] = a[j],a[i]
    return rev(a,i+1,j-1)

a = [5,7,3,2,6,1,5,9]
i = 2
j = 5
rev(a,i,j)
```

pallindrome string by recursion

```python
s = 'nitin'
dup = ''
j = len(s)-1
while j>=0:
    dup += s[j]
    j-=1
if dup==s:
    print('Pallindrome')
else:
    print('Not Pallindrome')
```

```python
def rev(s,j):
    if j<0:
        return ''
    return s[j] + rev(s,j-1)
s = 'nitin'
j = len(s)-1
if s == rev(s,j):
    print('Pallindrome')
else:
    print('Not pallindrome')
```

```python
s = 'nitin'
i = 0
j = len(s)-1
while i<=j:
    if s[i]==s[j]:
        i+=1
        j-=1
    else:
        print('Not pallindrome')
        break
else:print('Pallindrome')
```

```python
def reverse(s,left,right):
    if left>=right:
        return True
    if s[left]!=s[right]:
        return False
    return reverse(s,left+1,right-1)

s = 'malayalam'
left = 0
right = len(s)-1
reverse(s,left,right)
```

Fibonacci

```python
def fibonacci(n):
    if n==0 or n==1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)
n = 5
fibonacci(n)
```
