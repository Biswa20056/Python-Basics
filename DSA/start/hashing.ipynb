```python
s = 'azyxyyzaaaa'
q = ['d','a','y','x']
has_dict = [0]*123
for ch in s:
    has_dict[ord(ch)]+=1
for ch in q:
    if ord(ch)<97 or ord(ch)>122:
        print(0)
    else:
        print(has_dict[ord(ch)])
```

```python
s = 'azyxyyzaaaa'
q = ['d','a','y','x']
has_dict = [0]*26
for ch in s:
    ascii = ord(ch)
    index = ascii-97
    has_dict[index]+=1
for ch in q:
    if ord(ch)<97 or ord(ch)>122:
        print(0)
    else:
        ascii = ord(ch)
        index = ascii-97
        print(has_dict[index])
```
