```python
def linear(n,i=1):
    if i>n:
        return 
    print(i)
    linear(n,i+1)
i = 1
n = 10
linear(n,i)
```

```python
def linear(i,n):
    if n<0:
        return
    print(n)
    linear(i,n-1) 
i = 1
n = 5
linear(i,n)
```

print 1 to n by backtrackin(by not using i+1)

```python
def sample(n,i):
    if n<1:
        return 
    sample(n-1,i)
    print(n)
n = 3
i = 3
sample(n,i)
```

```python
def sample(n,i):
    if i>n:
        return
    sample(n,i+1)
    print(i)

n = 5
i = 1
sample(n,i)
```
